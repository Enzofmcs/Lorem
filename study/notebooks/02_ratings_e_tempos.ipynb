## O que o delta perde
Duas regiões com o mesmo delta podem ter níveis absolutos diferentes. Os mapas mostram mediana
(minutos) e P(AC≤30/60/120), usando a mesma escala 0–1 para probabilidades. Cinza é indisponível;
a tabela complementar registra eventos, censuras, participantes, problemas, contests/grupos e em risco.
